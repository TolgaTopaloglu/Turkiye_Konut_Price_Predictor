```python
X= df.drop("konut_fiyati_tl", axis=1)
y=df["konut_fiyati_tl"]
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
cat_col= X.select_dtypes(include=["object"]).columns
num_col= X.select_dtypes(exclude=["object"]).columns
print(cat_col)
print(num_col)
```

```python
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.preprocessing import StandardScaler, OneHotEncoder, FunctionTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.svm import SVR
from  sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.model_selection import cross_validate

log_transform= FunctionTransformer(func=np.log1p, validate=True, feature_names_out="one-to-one")

sayisal_pipeline=Pipeline([
    ("log_transform", log_transform),
    ("standard_scaler", StandardScaler())
])

preprocesser= ColumnTransformer(
    transformers=[
        ("sayisal", sayisal_pipeline, num_col),
        ("kategorik", OneHotEncoder(drop="first", handle_unknown="ignore"), cat_col)
    ], remainder="passthrough"
)


model_dic= {
    "Linear Regression": LinearRegression(),
    "Ridge": Ridge(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "AdaBoost": AdaBoostRegressor(random_state=42),
    "XGBoost": XGBRegressor(random_state=42),
    "LightGBM": LGBMRegressor(random_state=42, verbose=-1),
    "SVR": SVR(),
    "KNN": KNeighborsRegressor()
}
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
import mlflow
mlflow.set_tracking_uri("http://127.0.0.1:5000/")
mlflow.set_experiment("TOKI_price_regression")
```

```python
scoring= {
    "r2" : "r2",
    "mae" : "neg_mean_absolute_error",
    "mse" : "neg_mean_squared_error"
}

for name, model in model_dic.items():

    with mlflow.start_run(run_name=name):

        y_model= TransformedTargetRegressor(
            regressor=model, 
            func=np.log1p, 
            inverse_func=np.expm1
        )

        model_pipeline= Pipeline([
            ("preprocesser", preprocesser),
            ("model", y_model)
        ])

        scores= cross_validate(
            model_pipeline,
            X_train,
            y_train,
            cv=5, n_jobs=-1,  
            scoring=scoring
        )

        r2= scores["test_r2"].mean()
        mae= (-scores["test_mae"]).mean()
        mse= (-scores["test_mse"]).mean()

        mlflow.log_metrics({
            "r2_score" : r2,
            "mean_absolute_error" : mae,
            "mean_squared_error" : mse
        })

        print(name)
        print("r2", r2)
        print("MAE", mae)
        print("MSE", mse)
        print("-----------------------------\n\n")
        
```

en verimli model lightgbm 

```python
from sklearn.model_selection import GridSearchCV

ligtgbm_model= TransformedTargetRegressor(
    regressor=LGBMRegressor(random_state=42), 
    func=np.log1p, 
    inverse_func=np.expm1   
)

lightgbm_pipeline= Pipeline([
    ("preprocesser", preprocesser),
    ("model", ligtgbm_model)
])

lgbm_param={
    "model__regressor__n_estimators" : [100, 300 ,500],
    "model__regressor__learning_rate": [0.01, 0.05, 0.1],
    "model__regressor__max_depth" :  [3, 5, 7],
    "model__regressor__num_leaves"  : [20, 31, 40],
    "model__regressor__subsample" : [0.8, 1],
    "model__regressor__colsample_bytree" : [0.8, 1] #transformedtarget içindeki model olduğu için
}

lgbm_grid= GridSearchCV(estimator=lightgbm_pipeline, param_grid=lgbm_param, cv=5, n_jobs=-1, scoring="neg_mean_absolute_error", verbose=1 )

lgbm_grid.fit(X_train, y_train)
y_pred= lgbm_grid.predict(X_test)

r2= r2_score(y_test, y_pred)
mae= mean_absolute_error(y_test, y_pred)
mse= mean_squared_error(y_test, y_pred)

print("r2", r2)
print("MAE", mae)
print("MSE", mse)
```

```python
lgbm_grid.best_params_
```

```python
with mlflow.start_run(run_name="LGBM_Tunning_V1"):

    mlflow.log_params(lgbm_grid.best_params_)
    mlflow.log_metrics({
            "r2_score" : r2,
            "mean_absolute_error" : mae,
            "mean_squared_error" : mse
    })
```
